```python
from marker.converters.pdf import PdfConverter #before install marker-pdf pytorch has to be installed 
from marker.models import create_model_dict
from marker.output import text_from_rendered
import pandas as pd
#from paperscraper.arxiv import get_arxiv_papers_api
#from paperscraper.pdf import save_pdf_from_dump
from pathlib import Path
import os
import spacy
```

Two functions are defined: one to ensure that really only papers with an entry in the column journal and the other function is calling on the paperscraper-package to collect papers-

```python
def filter_non_empty_journal(papers):
    df = pd.DataFrame(papers)
    return df[
        df['journal'].notna() &
        (df['journal'].str.strip() != '') &
        (df['journal'].str.strip() != 'Unknown') &
        (df['journal'].str.strip() != 'N/A')
    ].reset_index(drop=True)

def collect_papers (query):
    papers = get_arxiv_papers_api(query, max_results = 1000)
    valid_papers = filter_non_empty_journal(papers)
    
    return valid_papers.drop_duplicates()
```

Calling the previously defined function to collect papers with "Nature", "ACM", or "IEEE" as their journal, saving the result as an Excel-file as well as printing the pandas dataframe.

```python
journal = collect_papers('jr:("Nature" OR "ACM" OR "IEEE")')
journal.to_excel("journal.xlsx")
journal
```

Get the name of every file within the folder to use for call and saving

```python
#extracting texts
converter = PdfConverter(
    artifact_dict=create_model_dict(),
)

path = "Journals"
dir_list = os.listdir(path)

for paper in dir_list:
    rendered = converter("Journals/" + paper)
    text, _, images = text_from_rendered(rendered)
    Path("Texts/" + str(paper[:-4]) + ".md").write_bytes(text.encode())
```

Next the pandas dataframe is enriched with the keywords of every paper to use them as basis for similar generated papers.

Problem: Arxiv erfasst keine Schlagwörter, sondern nur sehr grob die Kategorie

Mögliche Lösungen:
- Topic Modelling
    - Titel
    - Abstract
- direkt den Titel nehmen
- Titel oder Abstract bereinigen und auf Lemmata zurückführen, um somit allgemeinere Inhalte zu bekommen




```python
df = pd.read_excel("journal.xlsx")
```

```python
nlp = spacy.load("en_core_web_sm")
titles = df["title"].to_dict()
filtered_titles = {}

for id, title in titles.items():
    doc = nlp(title)
    filtered_title = [token.lemma_ for token in doc if (not token.is_stop ^ token.is_punct) ]
    filtered_titles[id] = filtered_title

filtered_titles
#and token.pos_ in {"NOUN", "VERB", "ADJ"}
```

```python
df["filtered_titles"] = pd.Series(filtered_titles)
df = df.drop('Unnamed: 0', axis=1)
df["List_entries"]=df.filtered_titles.map(len)
df.to_excel("journal_with_filtered_titles.xlsx")
```
